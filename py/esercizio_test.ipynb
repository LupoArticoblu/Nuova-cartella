```python
class Prodotto:
    def __init__(self, nome, quantita):
        self.nome = nome
        self.quantita = quantita
    
    def rifornisci(self,quantita_rif):
        self.quantita += quantita_rif
    
    def vendi(self,quantita_vend):
        if self.quantita >= quantita_vend:
            self.quantita -= quantita_vend
        else:
            return "Quantità insufficiente per la vendita"
    
    def descrizione(self):
        return f"Prodotto: {self.nome}, Quantità: {self.quantita}"
```

```python
#esercizio1

if __name__ == "__main__":
    prodotto1 = Prodotto("Biscotti", 100)
    print(prodotto1.descrizione())
    
    prodotto1.rifornisci(50)
    print(prodotto1.descrizione())
    
    prodotto1.vendi(30)
    print(prodotto1.descrizione())
    
    risultato_vendita = prodotto1.vendi(150)
    if risultato_vendita:
        print(risultato_vendita)
```
