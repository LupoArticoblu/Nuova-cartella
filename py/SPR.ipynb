```python
#creiamo un programma per giocare a SPR (Sasso, carta, forbici) contro il computer.
import random

#funzione per la scelta dell'utente
def get_user_choice():
    choice = input("Scegli sasso, carta o forbici (scrivi 'esci' per terminare il programma): ").lower()
    while choice not in ['sasso', 'carta', 'forbici', 'esci']:
        choice = input("Scelta non valida. Scegli sasso, carta o forbici: ").lower()
    return choice

#funzione per la scelta del computer
def get_computer_choice():
    return random.choice(['sasso', 'carta', 'forbici'])

#funzione per determinare il vincitore
def winneris(user_choice, computer_choice):
    if user_choice == computer_choice:
        return "Pareggio!"
    elif (user_choice == 'sasso' and computer_choice == 'forbici') or \
         (user_choice == 'carta' and computer_choice == 'sasso') or \
         (user_choice == 'forbici' and computer_choice == 'carta'):
        return "Hai vinto!"
    else:
        return "Hai perso!"

#ciclo di gioco
while True:
    #input per uscire dal gioco
    user_choice = get_user_choice()
    if user_choice == 'esci':
        break
    computer_choice = get_computer_choice()
    print("Hai scelto:", user_choice)
    print("Il computer ha scelto:", computer_choice)
    print(winneris(user_choice, computer_choice))
print("Grazie per aver giocato!")



```

# Scrivi una classe Python chiamata Prodotto.
 
Ogni prodotto deve avere:
 
un nome;,

una quantità disponibile in magazzino.

,
Il costruttore deve ricevere questi due valori e salvarli negli attributi dell’oggetto.
 
Aggiungi alla classe i seguenti metodi:
 
rifornisci(quantita), che aggiunge la quantità ricevuta alla quantità disponibile;,

vendi(quantita), che sottrae la quantità dalla disponibilità solo se la quantità disponibile è sufficiente;,

descrizione(), che restituisce una frase contenente il nome del prodotto e la quantità attualmente disponibile.

,
Il metodo vendi() deve restituire:
 
"Vendita eseguita" se la vendita è possibile;,

"Quantità insufficiente" se la quantità disponibile non è sufficiente.,


```python
class Prodotto:
    def __init__(self, nome, quantità_disponibile):
        self.nome = nome
        self.quantità_disponibile = quantità_disponibile

    #self ci và sempre perche sono proprietà della classe
    def rifornisci(self, quantità):
        self.quantità_disponibile += quantità

    def vendi(self, quantità):
        if self.quantità_disponibile >= quantità:
            self.quantità_disponibile -= quantità
            return "Vendita eseguita"
        return "Quantità insufficiente"

    def descrizione(self):
        return f"Prodotto: {self.nome}, Quantità disponibile: {self.quantità_disponibile}"
```

```python
prodotto1 = Prodotto("Penna", 20)

prodotto1.rifornisci(10)

print(prodotto1.vendi(15))
print(prodotto1.vendi(50))
print(prodotto1.descrizione())
```
